Write a Python program to find the stationary distribution $\pi = [\pi_0, \pi_1]$ satisfying $\pi P = \pi$ and $\pi_0 + \pi_1 = 1$.

(a) Rearrange the stationary-distribution equations into a system of linear equations $A\pi = b$.

(b) Solve the system using NumPy’s linear algebra function `numpy.linalg.solve()`.

(c) Verify numerically that $\pi P = \pi$ and that the probabilities sum to 1.

```python
import numpy as np 
p = [[0.75,0.25],
     [0.35,0.65]]
p = np.array(p)
n = np.shape(p)[0]
I = np.eye(n)

a = p.T - I
b = np.zeros(n)

# here a@pi = b

a[-1,:]=1
b[-1] = 1

pi = np.linalg.solve(a,b)

for index,element in enumerate(pi):
    print(f"pi{index} = {element}")

lhs = pi@p
rhs = pi

tot_prob = np.sum(p,axis = 1)
all_one = np.all(tot_prob==1)


if np.array_equal(lhs,rhs) and all_one:
    print("Valid Matrix")
else: 
    print("Non Valid")

```

$$
P =
\begin{bmatrix}

0.4 & 0.6 & 0 & 0 \\
0 & 0 & 0.7 & 0.3 \\
0 & 0 & 0.2 & 0.8
\end{bmatrix}
$$

Using Python and NetworkX:

(a) Construct a directed graph with an edge from state $i$ to state $j$ whenever $P_{ij} > 0$.

(b) Identify all communicating classes and display the states belonging to each class.

(c) State whether the Markov chain is irreducible, giving a reason based on the communicating classes.

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt


p = [[0.5,0.5,0,0],
     [0.4,0.6,0,0],
     [0,0,0.7,0.3],
     [0,0,0.2,0.8]]

p = np.array(p)
n = np.size(p,0)
g = nx.DiGraph()

g.add_nodes_from(range(0,n))

for i in range(n):
    for j in range(n):
        if(p[i][j]>0):
            g.add_edge(i,j,weight = p[i][j])

pos = nx.spring_layout(g,seed = 42)

nx.draw(g,pos,with_labels=True,node_color='red',edge_color='black',node_size=1200,
        connectionstyle='arc3,rad=0.2')

weights = nx.get_edge_attributes(g,'weight')
nx.draw_networkx_edge_labels(g,pos,edge_labels=weights,label_pos=0.3)


plt.show()

connection = list(nx.strongly_connected_components(g))
for index,item in enumerate(connection):
    print(f"class {index+1}-->{item}")

if (nx.is_strongly_connected(g)):
    print("This is irreducible")
else:
    print("This is reducible ")
```

10. A random variable $X$ follows a Binomial distribution with $n = 15$ and $p = 0.4$. Write a Python program to:

(a) Calculate $P(X = k)$ for all possible values of $k$.

(b) Calculate the mean and variance of the distribution.

(c) Plot the probability mass function using Matplotlib.

```python
import math
import matplotlib.pyplot as plt
n = 15
p = 0.4
q = 1-p
def get_probability(i):
    ans = (math.comb(n,i)) * (p**i) * (q**(n-i))
    return ans
k = list(range(0,n+1))
probability = list(map(get_probability,k))
print("probability = ",probability)

mean = n*p
variance = n*p*q
print(f'Mean = {mean} , Variance = {variance}')

#plotting 
plt.bar(k,probability,width=0.6,color='red')
plt.xlabel("K values")
plt.ylabel("probability")
plt.xticks(k)
plt.show()
```

19. Consider the transition probability matrix

$$P = \begin{bmatrix} 
1 & 0 & 0 \\ 
0.3 & 0.4 & 0.3 \\ 
0.2 & 0.5 & 0.3 
\end{bmatrix}$$

(a) Identify the absorbing state(s).
(b) Simulate 500 transitions starting from state 1.
(c) Estimate the proportion of simulations that reach an absorbing state within the simulated period.

```python
import numpy as np
p = [[1,0,0],
     [0.3,0.4,0.3],
     [0.2,0.5,0.3]]

p = np.array(p)

trap_states = []

n = np.size(p,axis=1)
for i in range(n):
    for j in range(n):
        if(i==j and p[i][j]==1):
            trap_states.append(i)

print(f"Absorbing States Are ",trap_states)

# b) Simulate 500 transition from state 1
rng = np.random.default_rng(42)
current_state = 1
tot_transition = 500
transition_states = []
for i in range(tot_transition+1):
    next_state = rng.choice([0,1,2],p=p[current_state])
    current_state=next_state
    transition_states.append(current_state)

print(f"500 Transition of State is ")
for it in transition_states:
    print(it,end=" ")

# c) number of simulation that reach to the trap state 

current_state = 1
num_simulation = 500
num_transition = 500
count=0
for _ in range(num_simulation):
    current_state = 1
    reached = False

    for _ in range(num_transition):
        next_state = rng.choice([0,1,2],p=p[current_state])
        current_state = next_state

        if current_state in trap_states:
            reached = True
            break

    if reached:
        count+=1
proportion = count/num_simulation
print(f"Simulation Reaching Absorbing State : {count}")
print(f"Estimated Proportion = {proportion}")
```

23. A single-server queue receives customers according to an exponential inter-arrival time with rate λ = 0.5.
Service times follow an exponential distribution with rate µ = 0.8. Simulate 200 customers. \
(a) Generate the arrival and service times. \
(b) Calculate the waiting time of each customer. \
(c) Find the average waiting time and server utilization

```python
import numpy as np
rng = np.random.default_rng()
lam = 0.5
mu = 0.8
n = 200
arrivalTime = rng.exponential(scale=(1/lam),size = n)
serviceTime = rng.exponential(scale=(1/mu),size=n)

arrivalTime = np.cumsum(arrivalTime)

waitingTime = np.zeros(n)
starttime = np.zeros(n)
departuretime = np.zeros(n)

for i in range(n):
    if(i==0):
        starttime[i]=arrivalTime[i]
    else:
        starttime[i] = max(arrivalTime[i],departuretime[i-1])


    departuretime[i] = starttime[i]+serviceTime[i]
    waitingTime[i] = starttime[i]-arrivalTime[i]

totTime = departuretime[-1]-arrivalTime[0]
totBusyTime = np.sum(serviceTime)

print(f"Utilization = {(totBusyTime/totTime)*100} %")

```

26. Write a Python program to perform basic statistical analysis on the list [12, 18, 15, 20, 25, 18, 14, 22, 16, 20]. \
(a) Convert the list into a NumPy array. \
(b) Calculate the mean, median, mode (where applicable), and standard deviation. \
(c) Sort the array and display the values in ascending order.

```python
import numpy as np
l = [12, 18, 15, 20, 25, 18, 14, 22, 16, 20, 25]
def findMode(arr):
    unique, frequency = np.unique(arr, return_counts=True)
    maxFreq = np.max(frequency)
    modes = set()
    for elem,freq in zip(unique,frequency):
        if(freq==maxFreq):
            modes.add(elem)
    return list(modes)

arr = np.array(l)
mean = np.mean(arr)
median = np.median(arr)
mode = findMode(arr)
std = np.std(arr)
print(f"Mean = {mean}")
print(f"Median = {median}")
print(f"mode = {mode}")
print(f"Standard Deviation = {std}")
arr = np.sort(arr)
for it in arr:
    print(it,end = " ")
print("\n")
```
